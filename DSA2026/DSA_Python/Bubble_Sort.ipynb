### Bubble Sort

```python
my_array = [64, 34, 25, 12, 22, 11,90,5]

n = len(my_array)
for i in range(n-1):
    for j in range(n-i-1):
        if my_array[j] > my_array[j+1]:
            my_array[j], my_array[j+1] = my_array[j+1], my_array[j]

print("Sorted Array :", my_array)
```

### Improved Bubble Sort

```python
my_array = [7,3,4,9,11,34]

n = len(my_array)
for i in range(n-1):
    swapped = False
    for j in range(n-i-1):
        if my_array[j] > my_array[j+1]:
            my_array[j], my_array[j+1] = my_array[j+1], my_array[j]
            swapped = True

    if not swapped:
        break

print("Sorted Array:", my_array)
            
```

### Time Complexity
O(n^2)
